```python
import pandas as pd
import json

# Caminho para seu arquivo
file_path = 'similarity_results.jsonl'

# Carrega cada linha do arquivo como um dicionário
data = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        item = json.loads(line)
        # Extrai campos desejados
        row = {
            'similarity_result': item['similarity_result'],
            'diagnosed_name': item['diagnosed']['name'],
            'diagnosed_given': item['diagnosed']['given'],
            'diagnosed_then': item['diagnosed']['then'],
            'candidate_name': item['candidate']['name'],
            'candidate_given': item['candidate']['given'],
            'candidate_then': item['candidate']['then'],
            'elapsed_time': item['elapsed_time'],
            'config': item['config'],
        }
        data.append(row)

# Cria o DataFrame
df = pd.DataFrame(data)

# Exemplo: mostra as primeiras linhas
df
```

```python
def extract_cfg_fields(row):
    cfg = row['config']  # já é um dict
    pw = cfg.get('parameter_weights', {})
    tw = cfg.get('tversky_weights', {})
    cw = cfg.get('conditional_weights', {})
    
    return pd.Series({
        'w_p1':    pw.get('p1'),
        'w_p10':   pw.get('p10'),
        'alpha':   tw.get('alpha'),
        'beta':    tw.get('beta'),
        'w_given': cw.get('given'),
        'w_when':  cw.get('when'),
        'w_then':  cw.get('then'),
    })

cfg_cols = df.apply(extract_cfg_fields, axis=1)
df_w = pd.concat([df, cfg_cols], axis=1)
df_w
```

```python
import plotly.express as px
import pandas as pd

df_plot = df_w.copy()

# 1) Mapeia nomes internos -> labels técnicos
label_map = {
    "cand_A_full_match_all_given_then": "baseline",
    "cand_B1_missing_p1_given": "missing-p1",
    "cand_B2_missing_p1_p2_given": "missing-p1p2",
    "cand_C1_extra_p12_given": "extra-p12",
    "cand_C2_extra_p12_p13_given": "extra-p12p13",
    "cand_D1_missing_p1_extra_p12_given": "missing-p1+extra-p12",
    "cand_D2_missing_p1_p2_extra_p12_p13_given": "missing-p1p2+extra-p12p13",
    "cand_E_parametric_only_p1_low": "param-p1-low",
}

df_plot["candidate_label"] = df_plot["candidate_name"].map(label_map)

# (opcional) define ordem dos candidatos no eixo Y
candidate_order = [
    "baseline",
    "missing-p1",
    "missing-p1p2",
    "extra-p12",
    "extra-p12p13",
    "missing-p1+extra-p12",
    "missing-p1p2+extra-p12p13",
    "param-p1-low",
]
df_plot["candidate_label"] = pd.Categorical(
    df_plot["candidate_label"],
    categories=candidate_order,
    ordered=True
)

# 2) cria rótulo da configuração (α, β)
df_plot["alpha_beta"] = df_plot.apply(
    lambda r: f"(α={r['alpha']:.1f}, β={r['beta']:.1f})",
    axis=1
)

config_order = sorted(df_plot["alpha_beta"].unique())
df_plot["alpha_beta"] = pd.Categorical(
    df_plot["alpha_beta"],
    categories=config_order,
    ordered=True
)

# 3) pivot: linhas = candidato (label), colunas = configuração
heat_df = df_plot.pivot_table(
    index="candidate_label",
    columns="alpha_beta",
    values="similarity_result"
)

fig = px.imshow(
    heat_df,
    text_auto=".2f",
    aspect="auto",
    labels={
        "x": "Penalty configuration (α, β)",
        "y": "Candidate",
        "color": "Similarity"
    },
    color_continuous_scale="Viridis"
)

fig.update_coloraxes(
    cmin=df_plot["similarity_result"].min(),
    cmax=df_plot["similarity_result"].max()
)

fig.update_layout(
    title=None,
    width=800,
    height=400,
    margin=dict(l=140, r=40, t=30, b=80),
    font=dict(size=12),
    xaxis=dict(tickangle=-10)
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False)

fig.show()
```

```python
import plotly.express as px
import pandas as pd

# rótulo legível para cada regime de pesos
df_plot = df_w.copy()
df_plot["penalty_regime"] = df_plot.apply(
    lambda r: f"T(α={r['alpha']:.1f}, β={r['beta']:.1f})",
    axis=1
)

regime_order = sorted(df_plot["penalty_regime"].unique())
df_plot["penalty_regime"] = pd.Categorical(
    df_plot["penalty_regime"],
    categories=regime_order,
    ordered=True
)

fig = px.line(
    df_plot,
    x="penalty_regime",
    y="similarity_result",
    color="candidate_name",
    markers=True,
    labels={
        "penalty_regime": "Penalty regime (α, β)",
        "similarity_result": "Similarity",
        "candidate_name": "Candidate"
    }
)

fig.update_traces(mode="lines+markers")

fig.update_layout(
    width=900,
    height=400,
    margin=dict(l=60, r=20, t=30, b=80),
    xaxis_tickangle=-30
)

fig.show()
```
