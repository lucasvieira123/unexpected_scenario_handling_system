```python
import pandas as pd
import json

# Caminho para seu arquivo
file_path = 'similarity_results.jsonl'

# Carrega cada linha do arquivo como um dicionário
data = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        item = json.loads(line)
        # Extrai campos desejados
        row = {
            'similarity_result': item['similarity_result'],
            'diagnosed_name': item['diagnosed']['name'],
            'diagnosed_given': item['diagnosed']['given'],
            'diagnosed_then': item['diagnosed']['then'],
            'candidate_name': item['candidate']['name'],
            'candidate_given': item['candidate']['given'],
            'candidate_then': item['candidate']['then'],
            'elapsed_time': item['elapsed_time'],
            'config': item['config'],
        }
        data.append(row)

# Cria o DataFrame
df = pd.DataFrame(data)

# Exemplo: mostra as primeiras linhas
df
```

```python
def extract_cfg_fields(row):
    cfg = row['config']  # já é um dict
    pw = cfg.get('parameter_weights', {})
    tw = cfg.get('tversky_weights', {})
    cw = cfg.get('conditional_weights', {})
    
    return pd.Series({
        'w_p1':    pw.get('p1'),
        'w_p10':   pw.get('p10'),
        'alpha':   tw.get('alpha'),
        'beta':    tw.get('beta'),
        'w_given': cw.get('given'),
        'w_when':  cw.get('when'),
        'w_then':  cw.get('then'),
    })

cfg_cols = df.apply(extract_cfg_fields, axis=1)
df_w = pd.concat([df, cfg_cols], axis=1)
df_w
```

```python
import plotly.express as px

# Mapeia nomes longos para rótulos mais curtos
label_map = {
    "cand_full_match_all_given_then": "Full match",
    "cand_error_only_p1_threshold_low": "p1 low",
    "cand_error_only_p1_threshold_high": "p1 high",
    "cand_error_only_p10_threshold_low": "p10 low",
    "cand_error_only_p10_threshold_high": "p10 high",
    "cand_errors_other_params_p2_low_p1_p10_correct": "other param low",
    "cand_errors_p1_p10_both_threshold_low": "p1 & p10 low",
    "cand_errors_p1_p10_both_threshold_high": "p1 & p10 high",
    "cand_errors_other_params_p2_high_p1_p10_correct": "other param high",
    "cand_errors_without_p1": "without p1",
    "cand_errors_without_p10": "without p10",
}

# 1) ordenar para as linhas ficarem “certinhas”
df_plot = df_w.sort_values(['candidate_name', 'w_p1']).copy()
df_plot["candidate_label"] = df_plot["candidate_name"].map(label_map).fillna(df_plot["candidate_name"])

# 2) gráfico de linhas: uma linha por candidato, x = peso de p1, y = similaridade
fig = px.line(
    df_plot,
    x='w_p1',
    y='similarity_result',
    color='candidate_label',
    markers=True,
    text=df_plot['similarity_result'].map(lambda v: f"{v:.3f}"),
    labels={
        'w_p1': 'Weight assigned to p1',   # <-- só p1 agora
        'similarity_result': 'Similarity',
        'candidate_label': 'Candidate'
    }
)

# 3) colocar os valores de similaridade acima dos pontos
fig.update_traces(textposition='top center')

# valores fixos desejados no eixo X (só p1 mudou)
x_vals   = [1.0, 2.0, 4.0, 8.0]
x_labels = [f"{v:.1f}" for v in x_vals]   # ou use f"w_p1 = {v:.1f}" se quiser

# 4) ajustar eixo X e legenda embaixo, horizontal
fig.update_layout(
    legend_title_text='Candidate:',
    legend=dict(
        orientation='h',
        x=0.5,
        xanchor='center',
        y=-0.25,
        yanchor='top'
    ),
    xaxis=dict(
        tickmode='array',
        tickvals=x_vals,    # posições no eixo (w_p1)
        ticktext=x_labels   # rótulos mostrando só o peso de p1
    )
)

fig.show()
```

```python
import plotly.express as px
import pandas as pd


# cria rótulo só com o peso de p1
df_plot["w_p1_str"] = df_plot["w_p1"].map(lambda v: f"{v:.1f}")

# ordem desejada no eixo x (pesos de p1)
ordem_pesos = ["1.0", "2.0", "4.0", "8.0"]
df_plot["w_p1_str"] = pd.Categorical(
    df_plot["w_p1_str"],
    categories=ordem_pesos,
    ordered=True
)

# >>> ordem desejada das ROWS (candidatos)
row_order = [
    "baseline",
    "low_p1",
    "high_p1",
    "low_p2",
    "high_p2",
    "without_p1",
    "without_p2",
    "extra_p12",
]

df_plot["candidate_label"] = pd.Categorical(
    df_plot["candidate_label"],
    categories=row_order,
    ordered=True
)

# pivot: linhas = candidatos, colunas = pesos de p1
heat_df = df_plot.pivot_table(
    index="candidate_label",
    columns="w_p1_str",
    values="similarity_result"
).reindex(row_order)   # garante a ordem das linhas

# heatmap
fig = px.imshow(
    heat_df,
    text_auto=".2f",
    aspect="auto",
    labels={
        "x": "p1 weight",
        "y": "Candidate",
        "color": "Similarity"
    },
    color_continuous_scale="Viridis",
)

fig.update_coloraxes(cmin=0.87, cmax=1.00)

fig.update_layout(
    title=None,
    xaxis_title="p1 weight",
    yaxis_title="Candidate",
    width=700,
    height=400,
    margin=dict(l=120, r=40, t=30, b=60),
    font=dict(size=12)
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False)

fig.show()
```
