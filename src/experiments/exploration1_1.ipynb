```python
df.drop(columns=['Desired When DNF',
       'Shared When DNF', 'Penality When', 'Global Similarity When',
       'Final Similarity When',"Desired Then DNF","Shared Then DNF","Penality Then","Global Similarity Then","Final Similarity Then"], inplace=True)
```

```python
def extrair_variaveis(expressao):
    variaveis = re.findall(r'(\w+)\s*(?:>=|<=|>|<|==|!=)', expressao)
    return ', '.join(variaveis)  # Opcional: juntar as variáveis em uma string separada por vírgula

# Aplicando a função para cada linha da coluna 'expressao' e armazenando na coluna 'vars'
df['desired_vars'] = df['Desired Given DNF'].apply(extrair_variaveis)
df['cadidate_vars'] = df['Shared Given DNF'].apply(extrair_variaveis)
df
```

```python
def calcular_intersecao(row):
    # Separar as variáveis em listas
    desired = set(row['desired_vars'].split(', '))
    candidate = set(row['cadidate_vars'].split(', '))
    
    # Calcular a interseção
    intersecao = desired & candidate
    return ', '.join(intersecao)  # Juntar as variáveis da interseção em uma string

# Aplicando a função para cada linha e criando uma nova coluna chamada 'intersection_vars'
df['intersection_vars'] = df.apply(calcular_intersecao, axis=1)
```

```python
#Função para contar quantas variáveis existem em uma string separada por ', '
def contar_variaveis(variaveis_str):
    if variaveis_str.strip() == '':
        return 0
    return len(variaveis_str.split(', '))

# Contando o número de variáveis em 'desired_vars', 'candidate_vars' e 'intersection_vars'
df['count_desired'] = df['desired_vars'].apply(contar_variaveis)
df['count_candidate'] = df['cadidate_vars'].apply(contar_variaveis)
df['count_intersection'] = df['intersection_vars'].apply(contar_variaveis)
# o numero de variaveis que o candidato não tem que esta presenta no desejado
df['complement_candidate'] = df['count_desired'] - df['count_intersection']
# o numero de variaveis que o desejado não tem que esta presente no candidato
df['extra_candidate'] = df['count_candidate'] - df['count_intersection']


df
```
