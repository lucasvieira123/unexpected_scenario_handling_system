```python
import json
import re

# Parâmetros usados nos blocos
param_given = ["p0", "p1", "p2", "p3", "p4"]
param_then = ["p6", "p7", "p8", "p9", "p10"]

def extract_x(expr, parametros):
    """Extrai o valor mínimo de cada parâmetro do texto (ex: p1 >= 50)"""
    xs = []
    for p in parametros:
        match = re.search(fr"{p} >= (\d+)", expr)
        if match:
            xs.append(int(match.group(1)))
        else:
            xs.append(0)  # ou None, se preferir
    return xs

def calcular_volume(xs):
    """Calcula o volume do hiperparalelepípedo dado os limites xs"""
    vol = 1
    for x in xs:
        v = max(0, 100 - x)
        vol *= v
    return vol

input_file = "similarity_results.jsonl"
output_file = "novo_arquivo_com_volumes.jsonl"

with open(input_file, "r", encoding="utf-8") as fin, open(output_file, "w", encoding="utf-8") as fout:
    for line in fin:
        obj = json.loads(line)
        
        # Diagnosed
        xs_given_diag = extract_x(obj["diagnosed"]["given"], param_given)
        xs_then_diag  = extract_x(obj["diagnosed"]["then"],  param_then)
        vol_given_diag = calcular_volume(xs_given_diag)
        vol_then_diag  = calcular_volume(xs_then_diag)
        
        # Candidate
        xs_given_cand = extract_x(obj["candidate"]["given"], param_given)
        xs_then_cand  = extract_x(obj["candidate"]["then"],  param_then)
        vol_given_cand = calcular_volume(xs_given_cand)
        vol_then_cand  = calcular_volume(xs_then_cand)
        
        # Novo objeto resumido
        new_obj = {
            "similarity": obj["similarity_result"],
            "diagnosed_given_volume": vol_given_diag,
            "diagnosed_then_volume": vol_then_diag,
            "candidate_given_volume": vol_given_cand,
            "candidate_then_volume": vol_then_cand
        }
        
        fout.write(json.dumps(new_obj) + "\n")

print(f"Arquivo '{output_file}' criado com os volumes e similaridades.")
```
