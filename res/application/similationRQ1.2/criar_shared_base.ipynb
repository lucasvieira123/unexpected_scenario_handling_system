```python
import itertools
import json

def powerset_nonempty(iterable):
    "Gera o powerset sem o conjunto vazio"
    s = list(iterable)
    return [list(combo) for r in range(1, len(s)+1) for combo in itertools.combinations(s, r)]

def generate_all_disjoint_scenarios(p_start=1, p_end=4, threshold=50,
                                    when_var="p11", action="action_x",
                                    name_prefix="diagnosed"):
    vars_all = [f"p{i}" for i in range(p_start, p_end + 1)]
    scenarios = []
    counter = 1

    given_subsets = powerset_nonempty(vars_all)
    then_subsets  = powerset_nonempty(vars_all)

    for given in given_subsets:
        for then in then_subsets:
            # Disjuntos: não podem ter variáveis em comum
            if set(given).isdisjoint(set(then)):
                given_str = " AND ".join(f"{v} >= {threshold}" for v in given)
                then_str  = " AND ".join(f"{v} >= {threshold}" for v in then)
                scenarios.append({
                    "name": f"{name_prefix}{counter}",
                    "given": given_str,
                    "when": f"{when_var} >= 0",
                    "do": action,
                    "then": then_str
                })
                counter += 1
    return scenarios

def save_scenarios_to_list_file(filename, scenarios):
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(scenarios, f, ensure_ascii=False, indent=2)

if __name__ == "__main__":
    scenarios = generate_all_disjoint_scenarios(
        p_start=1, p_end=5, threshold=50,
        when_var="p11", action="action_x",
        name_prefix="diagnosed"
    )
    save_scenarios_to_list_file("shared_scenarios.json", scenarios)
```
