```python
import pandas as pd
import json

# Caminho para seu arquivo
file_path = 'similarity_results.jsonl'

# Carrega cada linha do arquivo como um dicionário
data = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        item = json.loads(line)
        # Extrai campos desejados
        row = {
            'similarity_result': item['similarity_result'],
            'diagnosed_name': item['diagnosed']['name'],
            'diagnosed_given': item['diagnosed']['given'],
            'diagnosed_then': item['diagnosed']['then'],
            'candidate_name': item['candidate']['name'],
            'candidate_given': item['candidate']['given'],
            'candidate_then': item['candidate']['then'],
            'elapsed_time': item['elapsed_time'],
            'config': item['config'],
        }
        data.append(row)

# Cria o DataFrame
df = pd.DataFrame(data)

# Exemplo: mostra as primeiras linhas
df
```

```python
def extract_cfg_fields(row):
    cfg = row['config']  # já é um dict
    pw = cfg.get('parameter_weights', {})
    tw = cfg.get('tversky_weights', {})
    cw = cfg.get('conditional_weights', {})
    
    return pd.Series({
        'w_p1':    pw.get('p1'),
        'w_p10':   pw.get('p10'),
        'alpha':   tw.get('alpha'),
        'beta':    tw.get('beta'),
        'w_given': cw.get('given'),
        'w_when':  cw.get('when'),
        'w_then':  cw.get('then'),
    })

cfg_cols = df.apply(extract_cfg_fields, axis=1)
df_w = pd.concat([df, cfg_cols], axis=1)
df_w
```

```python
import plotly.express as px
import pandas as pd

df_plot = df_w.copy()

# 1) Mapeia nomes internos -> labels técnicos (RQ3.3)
label_map = {
    "cand_baseline":          "baseline",
    "cand_given_low":         "given-low",
    "cand_given_high":        "given-high",
    "cand_then_low":          "then-low",
    "cand_then_high":         "then-high",
    "cand_given_then_low":    "given+then-low",
    "cand_given_then_high":   "given+then-high",
    "cand_when_high":         "when-high",
}

df_plot["candidate_label"] = df_plot["candidate_name"].map(label_map).fillna(df_plot["candidate_name"])

# Ordem dos candidatos no eixo Y (pode ajustar se quiser outra ordem)
candidate_order = [
    "baseline",
    "given-low",
    "given-high",
    "then-low",
    "then-high",
    "given+then-low",
    "given+then-high",
    "when-high",
]
df_plot["candidate_label"] = pd.Categorical(
    df_plot["candidate_label"],
    categories=candidate_order,
    ordered=True
)

# 2) Cria rótulo da configuração de pesos de cláusula (w_given, w_when, w_then)
#    Mapeando para as 5 configs definidas na tabela (C1–C5)
config_map = {
    (1.0, 1.0, 1.0): "C1 (1,1,1)",
    (3.0, 1.0, 1.0): "C2 (3,1,1)",
    (1.0, 1.0, 3.0): "C3 (1,1,3)",
    (1.0, 3.0, 1.0): "C4 (1,3,1)",
    (2.0, 1.0, 2.0): "C5 (2,1,2)",
}

def label_clause_config(row):
    key = (row["w_given"], row["w_when"], row["w_then"])
    return config_map.get(
        key,
        f"({row['w_given']:.1g},{row['w_when']:.1g},{row['w_then']:.1g})"
    )

df_plot["clause_config"] = df_plot.apply(label_clause_config, axis=1)

# Ordem das configurações no eixo X
config_order = ["C1 (1,1,1)", "C2 (3,1,1)", "C3 (1,1,3)", "C4 (1,3,1)", "C5 (2,1,2)"]
df_plot["clause_config"] = pd.Categorical(
    df_plot["clause_config"],
    categories=config_order,
    ordered=True
)

# 3) pivot: linhas = candidato, colunas = configuração (pesos de cláusula)
heat_df = df_plot.pivot_table(
    index="candidate_label",
    columns="clause_config",
    values="similarity_result"
)

fig = px.imshow(
    heat_df,
    text_auto=".2f",
    aspect="auto",
    labels={
        "x": "Clause-weight configuration (w_given, w_when, w_then)",
        "y": "Candidate",
        "color": "Similarity"
    },
    color_continuous_scale="Viridis"
)

# Mantém a escala de cores consistente
fig.update_coloraxes(
    cmin=df_plot["similarity_result"].min(),
    cmax=df_plot["similarity_result"].max()
)

fig.update_layout(
    title=None,
    width=800,
    height=450,
    margin=dict(l=150, r=40, t=30, b=80),
    font=dict(size=12),
)

fig.update_xaxes(showgrid=False, tickangle=-15)
fig.update_yaxes(showgrid=False)

fig.show()
```
